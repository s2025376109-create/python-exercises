```
def string_operations(s):
  reversed_str = s[::-1]
  every_second = s[::2]
  return reversed_str, every_second


# Example usage
rev, sec = string_operations("Artificial Intelligence")
print("Reversed:", rev)
print("Every 2nd char:", sec)
```

```
def even_squares(numbers):
  return [x**2 for x in numbers if x % 2 == 0]


# Example usage
nums = [1, 2, 3, 4, 5, 6, 7, 8]
print("Squares of evens:", even_squares(nums))
```

```
def char_frequency(s):
  freq = {}
  for char in s:
    freq[char] = freq.get(char, 0) + 1
  return freq


# Example usage
print("Character frequency:", char_frequency("programming"))
```

```
def lists_to_dict(keys, values):
  return dict(zip(keys, values))


# Example usage
k = ["name", "course", "marks"]
v = ["Ali", "AI 271", 95]
print("Resulting dictionary:", lists_to_dict(k, v))
```

```
my_list = [10, 20, 30, 40]
iterator = iter(my_list)

while True:
  try:
    item = next(iterator)
    print(item)
  except StopIteration:
    print("Reached the end of the list (StopIteration handled).")
    break
```

```
def is_palindrome(s):
  # Optional: normalize case and spaces if needed: s = s.lower().replace(" ", "")
  return s == s[::-1]


# Example usage
print("Is 'madam' a palindrome?:", is_palindrome("madam"))
print("Is 'python' a palindrome?:", is_palindrome("python"))
```

```
def second_largest(numbers):
  if len(numbers) < 2:
    return None

  largest = second = float("-inf")
  for num in numbers:
    if num > largest:
      second = largest
      largest = num
    elif largest > num > second:
      second = num

  return second if second != float("-inf") else None


# Example usage
print("Second largest:", second_largest([12, 35, 1, 10, 34, 1]))
```
